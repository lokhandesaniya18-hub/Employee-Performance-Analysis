# Employee Performance Prediction

## Objective
This notebook loads the trained machine learning pipeline and predicts employee performance ratings.

## Dataset
The notebook uses the processed INX Future Inc. employee dataset.

## Prediction Workflow
1. Load the saved model pipeline.
2. Load the processed dataset.
3. Separate input features and target.
4. Generate predictions on the test data.
5. Compare actual and predicted ratings.

## Evaluation
The saved model achieved 93.75% accuracy on the test split.

```python
import joblib
from pathlib import Path

MODEL_PATH = Path("models/final_employee_performance_model.pkl")

model = joblib.load(MODEL_PATH)

print("Model loaded successfully!")
print("Model type:", type(model).__name__)
```

```python
import pandas as pd
from pathlib import Path
from sklearn.metrics import accuracy_score, classification_report

DATA_PATH = Path("../../data/processed/INX_employee_performance_processed.csv")

df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
df = pd.read_csv(DATA_PATH)

# Separate input features and actual target
X = df.drop(columns=["PerformanceRating", "EmpNumber"])
y = df["PerformanceRating"]

# Use the same test split as model training
from sklearn.model_selection import train_test_split

_, X_test, _, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Generate predictions
y_pred = model.predict(X_test)

print("Actual ratings:", y_test.head(10).tolist())
print("Predicted ratings:", y_pred[:10].tolist())
print("Test Accuracy:", accuracy_score(y_test, y_pred))
```

```python
results = pd.DataFrame({
    "Actual Rating": y_test.values,
    "Predicted Rating": y_pred
})

results["Correct Prediction"] = (
    results["Actual Rating"] == results["Predicted Rating"]
)

results.index.name = "Test Record"

display(results.head(10))
```
